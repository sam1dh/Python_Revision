### **1. Introduction to File Handling**

File handling allows a Python program to read data from a file or write data to a file. This is crucial because variables and data structures (like lists and dictionaries) only store data in the computer's temporary memory (RAM). When the program ends, that data is lost. Files allow for **data persistence** (saving data to a hard drive).

The basic lifecycle of a file operation in Python is always:

1. **Open** the file.
2. **Process** the file (Read or Write).
3. **Close** the file (to free up system resources).

---

### **2. Opening a File**

Python uses the built-in `open()` function to open a file. It returns a file object.

**Syntax:**

```python
file = open("filename.txt", "mode")

```

#### **File Modes (Crucial to memorize):**

| Mode | Description | File Pointer Position | Overwrites Existing? | Creates New File? |
| --- | --- | --- | --- | --- |
| **`'r'`** | **Read** (Default). Opens for reading. Error if file doesn't exist. | Beginning | No | No |
| **`'w'`** | **Write**. Opens for writing. | Beginning | **Yes** (Truncates) | Yes |
| **`'a'`** | **Append**. Opens for writing. Adds new data to the end. | End | No | Yes |
| **`'x'`** | **Exclusive creation**. Creates a new file. Fails if file exists. | Beginning | No | Yes |

**Modifiers (can be combined with modes above, e.g., `'rb'`, `'w+'`):**

* **`'t'`**: Text mode (Default). Handles strings.
* **`'b'`**: Binary mode. Handles bytes (used for images, PDFs, etc.).
* **`'+'`**: Updates mode. Opens for both reading and writing (e.g., `'r+'` reads and writes, `'w+'` writes and reads but overwrites the file first).

---

### **3. The "Pythonic" Way to Open Files (Context Managers)**

While you can use `file = open(...)` and `file.close()`, it is easy to forget to close the file, which can lead to memory leaks or corrupted files if an error crashes your program midway.

**Best Practice:** Always use the `with` statement. It automatically closes the file for you, even if an exception (error) occurs within the block.

```python
# The recommended way
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
# The file is automatically closed here, outside the indented block

```

---

### **4. Reading from a File**

Assume we have a file opened in read mode: `file = open("data.txt", "r")`

* **`file.read(size)`**: Reads the entire file as a single string. If `size` (in bytes) is specified, it reads only that much.
* **`file.readline()`**: Reads a single line from the file (including the newline character `\n`).
* **`file.readlines()`**: Reads all lines and returns them as a **list of strings**, where each item is a line.

**Iterating line-by-line (Most memory efficient for large files):**

```python
with open("large_file.txt", "r") as file:
    for line in file:
        print(line.strip()) # .strip() removes the trailing newline character

```

---

### **5. Writing to a File**

Assume we have a file opened in write or append mode: `file = open("output.txt", "w")`

* **`file.write(string)`**: Writes a string to the file. It does *not* automatically add a newline, so you must include `\n` if you want one.
* **`file.writelines(list)`**: Writes a list of strings to the file. Again, it does not add newlines automatically.

```python
lines_to_write = ["First line\n", "Second line\n", "Third line\n"]

with open("output.txt", "w") as file:
    file.write("Hello World!\n")
    file.writelines(lines_to_write)

```

---

### **6. File Pointer Manipulation**

The "file pointer" is like a cursor that tracks where you are currently reading or writing in the file.

* **`file.tell()`**: Returns the current position of the file pointer (in bytes from the beginning).
* **`file.seek(offset, whence)`**: Moves the file pointer to a specific position.
* `offset`: Number of bytes to move.
* `whence`: Reference point (0 = beginning [default], 1 = current position, 2 = end). Note: In Python 3 text files, `whence` only supports 0 (beginning) unless the offset is 0.



```python
with open("example.txt", "r") as file:
    print(file.read(5))  # Reads first 5 characters
    print(file.tell())   # Outputs: 5
    file.seek(0)         # Moves pointer back to the very beginning
    print(file.read(5))  # Reads the first 5 characters again

```

### **7. Deleting Files**

File handling doesn't include file deletion directly. To delete or rename files, you need to import the `os` module.

```python
import os

if os.path.exists("file_to_delete.txt"):
    os.remove("file_to_delete.txt")
else:
    print("The file does not exist")

```
