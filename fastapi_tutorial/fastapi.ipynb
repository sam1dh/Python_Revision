# FastAPI Notes (Till Path Parameters & Query Parameters)

# 1. What is FastAPI?

FastAPI is a modern Python framework used to build APIs quickly and efficiently.

Main advantages:

* High performance (uses ASGI)
* Automatic API documentation
* Type validation
* Easy request handling
* Supports async programming

Installation:

```bash
pip install fastapi
pip install uvicorn
```

Run the application:

```bash
uvicorn filename:app --reload
```

Example:

```bash
uvicorn main:app --reload
```

Where:

* `main` → file name
* `app` → FastAPI object
* `--reload` → automatically reloads server after changes

---

# 2. Creating FastAPI Object

Code:

```python
from fastapi import FastAPI

app = FastAPI()
```

Explanation:

`FastAPI()` creates an application instance.

Everything is attached to this object.

Think of it as:

```text
app
│
├── Routes
├── Validation
├── Documentation
└── Middleware
```

# 3. Creating Routes

Routes define endpoints.

Syntax:

```python
@app.method("URL")
def function():
```

Example:

```python
@app.get("/")
def hello():
    return {"Hello":"World"}
```

Output:

```json
{
   "Hello":"World"
}
```

Common methods:

| Method | Purpose              |
| ------ | -------------------- |
| GET    | Retrieve data        |
| POST   | Add data             |
| PUT    | Update complete data |
| PATCH  | Partial update       |
| DELETE | Remove data          |

# 5. Path Parameters ⭐⭐⭐⭐⭐

Most important topic.

Path parameters are values passed inside URL paths.

Example:

```python
@app.get("/patient/{id}")
```

URL:

```text
http://localhost:8000/patient/1
```

FastAPI extracts:

```python
id="1"
```

Function:

```python
def get_patient_id(id:str):
```

Flow:

```text
User request
      ↓
/patient/1
      ↓
FastAPI extracts value
      ↓
id="1"
      ↓
Function executes
```

# HTTPException ⭐⭐⭐⭐⭐

Used for custom errors.

Syntax:

```python
raise HTTPException(
    status_code=404,
    detail="Patient not found"
)
```

Example:

```python
if id not in data:
    raise HTTPException(
        status_code=404,
        detail="Patient not found"
    )
```

Output:

```json
{
    "detail":"Patient not found"
}
```

Common status codes:

| Code | Meaning               |
| ---- | --------------------- |
| 200  | Success               |
| 201  | Created               |
| 400  | Bad request           |
| 401  | Unauthorized          |
| 403  | Forbidden             |
| 404  | Not found             |

# 6. Query Parameters ⭐⭐⭐⭐⭐

Query parameters are values passed after `?`

Example:

```text
http://localhost:8000/sort?sort_by=age&order=asc
```

FastAPI extracts:

```python
sort_by="age"

order="asc"
```

Function:

```python
@app.get("/sort")
def sort_patient(
    sort_by:str,
    order:str
):
```

Flow:

```text
Request
    ↓

/sort?sort_by=age&order=asc

    ↓

sort_by="age"
order="asc"

    ↓

Function executes
```

---

# Query() Function

Code:

```python
sort_by:str=Query(
    ...,
    description="Sort by age"
)
```

Purpose:

* Validation
* Documentation
* Default values

Example:

```python
order:str=Query(
    "asc",
    description="sort order"
)
```

Default:

```python
order="asc"
```

If user writes:

```text
/sort?sort_by=age
```

Output automatically becomes:

```python
order="asc"
```

---

# MOST IMPORTANT CONCEPT #3

Difference between Path and Query Parameters

| Path Parameter      | Query Parameter    |
| ------------------- | ------------------ |
| Required by default | Optional usually   |
| Part of URL path    | Comes after ?      |
| Identifies resource | Filters/sorts data |

Examples:

Path:

```text
/patient/1
```

Query:

```text
/sort?sort_by=age
```

Real-world examples:

Get specific user:

```text
/users/10
```

Search users:

```text
/users?city=Mumbai
```

Filter products:

```text
/products?price=1000
```
