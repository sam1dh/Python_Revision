---

## 1. Introduction & Creation

* **Definition**: Ordered, mutable collections of items.
* **Creation**: Use square brackets `[]` or the `list()` constructor.
```python
my_list = [10, "Python", 3.14]

```



---

## 2. Accessing & Modifying

* **Accessing**: Use zero-based indexing (e.g., `my_list[0]`) or negative indexing for the end (e.g., `my_list[-1]`).
* **Modifying**: Assign a new value directly to an index: `my_list[1] = "Changed"`.

---

## 3. List Methods

* **`append(x)`**: Adds an item to the end.
* **`extend(iterable)`**: Adds all elements of another list.
* **`insert(i, x)`**: Inserts item at a specific index.
* **`remove(x)`**: Removes the first occurrence of a value.
* **`pop(i)`**: Removes and returns the item at index `i` (defaults to the last item).

---

## 4. Slicing Lists

* **Syntax**: `list[start:stop:step]`.
* **Example**: `my_list[1:4]` gets elements from index 1 to 3.
* **Reverse**: `my_list[::-1]`.

---

## 5. Iteration & Comprehensions

* **Iterating**: Use a `for` loop to go through elements.
```python
for item in my_list:
    print(item)

```


* **List Comprehensions**: Concise way to create lists.
```python
squares = [x**2 for x in range(5)] # [0, 1, 4, 9, 16]

```



---

## 6. Advanced Concepts

* **Nested Lists**: Lists containing other lists (e.g., `matrix = [[1, 2], [3, 4]]`).
* **Common Errors**:
* `IndexError`: Accessing an index that doesn't exist.
* `ValueError`: Trying to remove an item that is not in the list.
