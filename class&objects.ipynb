### **1. The Core Concept: Object-Oriented Programming (OOP)**

Python is an **Object-Oriented Programming** language. This means it uses "objects" to represent data and methods.

Think of OOP as a way to structure your code by mimicking the real world. Instead of writing isolated variables and functions, you bundle them together into logical packages.

* **Class:** A **blueprint** or template for creating objects. It defines the structure and behavior.
* **Object:** An **instance** of a class. It is the actual entity created from the blueprint, filled with real data.

> 💡 **The Architecture Analogy:**
> Think of a **Class** as the blueprint for a house. The blueprint itself isn't a house—you can't live in it. An **Object** is the actual physical house built using that blueprint. You can build 50 identical houses (objects) from a single blueprint (class), but each house can have different colors or furniture inside.

---

### **2. Defining a Class and Creating an Object**

In Python, you define a class using the `class` keyword. By convention, class names use **PascalCase** (CapitalizeTheFirstLetterOfEachWord).

```python
# Defining the Blueprint (Class)
class Car:
    pass  # 'pass' is just a placeholder for now

# Creating an Instance (Object)
my_car = Car()
print(type(my_car))  # Outputs: <class '__main__.Car'>

```

---

### **3. Attributes and Methods**

A class primarily consists of two things:

1. **Attributes:** Variables inside a class (what the object *knows* / its characteristics).
2. **Methods:** Functions inside a class (what the object *does* / its actions).

#### **The `__init__()` Method (The Constructor)**

When you create a new object, Python automatically runs a special method called `__init__()`. It is used to initialize the object's attributes.

#### **What is `self`?**

The `self` parameter represents the **specific object** you are currently creating or working with. It allows the class to access its own attributes and methods. You must include it as the first parameter in almost all class methods, but Python handles passing it automatically when you call the method.

---

### **4. A Complete Example**

Here is a practical example putting attributes, methods, `__init__`, and `self` together:

```python
class Dog:
    # The Constructor
    def __init__(self, name, breed):
        self.name = name    # Instance Attribute
        self.breed = breed  # Instance Attribute

    # A Method (Action)
    def bark(self):
        return f"{self.name} says Woof!"

    # Another Method using attributes
    def get_details(self):
        return f"{self.name} is a {self.breed}."


# ---- Creating Objects (Instantiation) ----

dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Max", "German Shepherd")

# ---- Accessing Attributes and Methods ----

# Accessing attributes
print(dog1.name)   # Outputs: Buddy
print(dog2.breed)  # Outputs: German Shepherd

# Calling methods
print(dog1.bark())         # Outputs: Buddy says Woof!
print(dog2.get_details())  # Outputs: Max is a German Shepherd.

```

---

### **5. Class Attributes vs. Instance Attributes**

It is important to understand where your data lives:

* **Instance Attributes:** Defined inside `__init__`. They are unique to each specific object (e.g., `dog1` has a different name than `dog2`).
* **Class Attributes:** Defined directly inside the class, outside any methods. They are **shared** by all objects of that class.

```python
class Circle:
    pi = 3.14159  # Class Attribute (Shared by all circles)

    def __init__(self, radius):
        self.radius = radius  # Instance Attribute (Unique to each circle)


circle1 = Circle(5)
circle2 = Circle(10)

print(circle1.pi)      # Outputs: 3.14159
print(circle2.pi)      # Outputs: 3.14159
print(circle1.radius)  # Outputs: 5

```
