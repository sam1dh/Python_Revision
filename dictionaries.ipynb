### Python Dictionary Fast Revision

Dictionaries are **unordered** (in older versions), **mutable**, and **indexed** by keys. They store data in `key:value` pairs.

---

## 1. Creation & Basics

* **Syntax**: Uses curly braces `{}` with keys and values separated by colons.
```python
# Creation
student = {"name": "Mayank", "age": 21, "course": "CS"}

# Empty dictionary
empty_dict = {}
# Using constructor
another_dict = dict(brand="Ford", model="Mustang")

```



---

## 2. Accessing & Modifying

* **Accessing**: Use the key inside square brackets or the `.get()` method.
* `student["name"]` $\rightarrow$ Returns "Mayank".
* `student.get("age")` $\rightarrow$ Returns 21 (safer, returns `None` if key is missing).


* **Modifying**: Use the key to update an existing value or add a new pair.
* `student["age"] = 22` (Updates value).
* `student["city"] = "Pune"` (Adds new key-value pair).



---

## 3. Essential Dictionary Methods

| Method | Description |
| --- | --- |
| **`.keys()`** | Returns a list of all keys in the dictionary. |
| **`.values()`** | Returns a list of all values in the dictionary. |
| **`.items()`** | Returns a list of tuples (key, value) for each pair. |
| **`.update({k:v})`** | Updates the dictionary with the specified key-value pairs. |
| **`.pop("key")`** | Removes the specified key and returns its value. |
| **`.clear()`** | Removes all elements from the dictionary. |

---

## 4. Iterating Over Dictionaries

You can loop through a dictionary in several ways:

```python
# Loop through keys
for key in student:
    print(key)

# Loop through values
for val in student.values():
    print(val)

# Loop through both (Commonly used)
for key, value in student.items():
    print(f"{key}: {value}")

```

---

## 5. Dictionary Comprehension

Just like lists, you can create dictionaries concisely:

```python
# Create a dictionary of squares
squares = {x: x**2 for x in range(5)} 
# Result: {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}

```

---

## 6. Key Rules

* **Keys must be unique**: Duplicate keys will overwrite the existing value.
* **Keys must be immutable**: You can use strings, numbers, or tuples as keys, but **not lists** (because lists can change).
