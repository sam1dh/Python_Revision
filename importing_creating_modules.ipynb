### Python Modules & Packages: Fast Revision

In Python, **Modules** are single `.py` files containing code, while **Packages** are directories containing multiple modules.

---

## 1. Modules

A module is simply a file containing Python definitions and statements.

### Creating a Module:

Save a file as `mymodule.py`:

```python
# mymodule.py
def greet(name):
    return f"Hello, {name}!"

version = "1.0"

```

### Importing a Module:

You can bring code into your main script using different `import` styles:

| Style | Syntax | Usage |
| --- | --- | --- |
| **Direct Import** | `import mymodule` | `mymodule.greet("Mayank")` |
| **Alias** | `import mymodule as mm` | `mm.greet("Mayank")` |
| **Specific Item** | `from mymodule import greet` | `greet("Mayank")` |
| **Import All** | `from mymodule import *` | Imports everything (not recommended) |

---

## 2. Packages

A package is a collection of modules organized in a folder hierarchy.

### Creating a Package:

To turn a folder into a package, it must contain a file named `__init__.py` (it can be empty).

**Folder Structure:**

```text
my_project/
│
├── main.py             # Your entry point
└── my_package/         # The Package
    ├── __init__.py     # Makes folder a package
    ├── module1.py      # Sub-module
    └── module2.py      # Sub-module

```

### Importing from a Package:

```python
# From main.py
from my_package import module1
from my_package.module2 import some_function

```

---

## 3. The `__name__ == "__main__"` Block

This is used to prevent code from running automatically when a module is imported.

```python
def main():
    print("This runs only if the file is executed directly.")

if __name__ == "__main__":
    main()

```

## 5. Key Concepts

* **`sys.path`**: A list of strings that specifies the search path for modules.
* **Namespace**: A container where names (variables, functions) are mapped to objects to avoid naming conflicts.
* **`__init__.py`**: In newer Python versions (3.3+), this file is optional for "namespace packages," but it's still best practice to include it for regular packages.

In Python, every file is treated as a **module**.
The special variable `__name__` helps Python identify:

* whether the file is being run directly
  OR
* whether it is being imported into another file.

This line:

```python
if __name__ == "__main__":
```

is one of the most important concepts in real-world Python development.

# Step-by-Step Understanding

## 1. What is `__name__`?

`__name__` is a **special built-in variable** automatically created by Python.

Its value changes depending on how the file is used.

---

# Case 1: File is Run Directly

Suppose file name is:

```python
app.py
```

Code:

```python
print(__name__)
```

If you run:

```bash
python app.py
```

Output:

```python
__main__
```

So when Python executes a file directly:

```python
__name__ = "__main__"
```

---

# Case 2: File is Imported

Suppose:

## file1.py

```python
print(__name__)
```

## file2.py

```python
import file1
```

Now run:

```bash
python file2.py
```

Output:

```python
file1
```

Because now:

* `file1.py` is NOT the main file
* it is being imported

So:

```python
__name__ = "file1"
```

(the module name)

---

# Why This Matters

Sometimes you want:

* some code to run ONLY when the file is executed directly
* but NOT run when imported elsewhere

That is why we use:

```python
if __name__ == "__main__":
```

---

# Real Structure

```python
def main():
    print("Program started")

if __name__ == "__main__":
    main()
```

---

# What Happens Here?

## If file is run directly:

```bash
python app.py
```

Then:

```python
__name__ == "__main__"
```

Condition becomes TRUE.

So:

```python
main()
```

runs.

Output:

```python
Program started
```

---

# If Imported

Suppose:

## app.py

```python
def main():
    print("Program started")

if __name__ == "__main__":
    main()
```

## test.py

```python
import app
```

Now run:

```bash
python test.py
```

What happens?

Python imports `app.py`.

Now:

```python
__name__ = "app"
```

Condition:

```python
if "app" == "__main__":
```

FALSE.

So `main()` does NOT run automatically.

This prevents unwanted execution.

---

# Why Developers Use This

This is extremely common in:

* backend development
* scripts

# Without This Block (Problem)

Example:

## math_utils.py

```python
print("Loading module...")

def add(a, b):
    return a + b

print(add(2, 3))
```

Now another file imports it:

```python
import math_utils
```

Output:

```python
Loading module...
5
```

Even though you only wanted the function!

This is bad design.

---

# Correct Version

```python
def add(a, b):
    return a + b

def main():
    print("Loading module...")
    print(add(2, 3))

if __name__ == "__main__":
    main()
```

Now importing the module:

* imports functions only
* does NOT execute testing/demo code

Much cleaner.

# Industry Usage

* reusable code
  from
* executable code

# One More Important Concept

Python executes code from top to bottom.

So:

```python
print("Hello")
```

runs immediately during import too.

Functions only run when called.

That is why we protect execution code using:

```python
if __name__ == "__main__":
```

# Real-World Example (API)

Using FastAPI:

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello"}

if __name__ == "__main__":
    import uvicorn
    uvicorn.run(app)
```

This ensures:

* server starts only when file executed directly
* not when imported elsewhere
