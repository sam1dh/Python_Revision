### Python Modules & Packages: Fast Revision

In Python, **Modules** are single `.py` files containing code, while **Packages** are directories containing multiple modules.

---

## 1. Modules

A module is simply a file containing Python definitions and statements.

### Creating a Module:

Save a file as `mymodule.py`:

```python
# mymodule.py
def greet(name):
    return f"Hello, {name}!"

version = "1.0"

```

### Importing a Module:

You can bring code into your main script using different `import` styles:

| Style | Syntax | Usage |
| --- | --- | --- |
| **Direct Import** | `import mymodule` | `mymodule.greet("Mayank")` |
| **Alias** | `import mymodule as mm` | `mm.greet("Mayank")` |
| **Specific Item** | `from mymodule import greet` | `greet("Mayank")` |
| **Import All** | `from mymodule import *` | Imports everything (not recommended) |

---

## 2. Packages

A package is a collection of modules organized in a folder hierarchy.

### Creating a Package:

To turn a folder into a package, it must contain a file named `__init__.py` (it can be empty).

**Folder Structure:**

```text
my_project/
│
├── main.py             # Your entry point
└── my_package/         # The Package
    ├── __init__.py     # Makes folder a package
    ├── module1.py      # Sub-module
    └── module2.py      # Sub-module

```

### Importing from a Package:

```python
# From main.py
from my_package import module1
from my_package.module2 import some_function

```

---

## 3. The `__name__ == "__main__"` Block

This is used to prevent code from running automatically when a module is imported.

```python
def main():
    print("This runs only if the file is executed directly.")

if __name__ == "__main__":
    main()

```

## 4. Built-in Modules & `pip`

Python comes with a "Standard Library" of built-in modules:

* `math`: Math functions (`math.sqrt()`).
* `os`: Operating system interactions.
* `sys`: System-specific parameters.
* `random`: Generate random numbers.

**Installing External Packages:**
Use `pip` (Python Package Index) in your terminal:

```bash
pip install requests

```

---

## 5. Key Concepts

* **`sys.path`**: A list of strings that specifies the search path for modules.
* **Namespace**: A container where names (variables, functions) are mapped to objects to avoid naming conflicts.
* **`__init__.py`**: In newer Python versions (3.3+), this file is optional for "namespace packages," but it's still best practice to include it for regular packages.
