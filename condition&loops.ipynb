## 1. Conditional Statements

Used to execute code blocks based on specific conditions. Python uses **indentation** to define the scope of these blocks.

* **`if`**: Executes code if the condition is true.
* **`elif`** (short for else if): Checks another condition if the previous ones were false.
* **`else`**: Executes if all previous conditions were false.

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
else:
    print("Grade: C")

```

**Short Hand (Ternary Operator):**
`result = "Pass" if score >= 40 else "Fail"`

---

## 2. For Loops

Used for iterating over a sequence (list, tuple, dictionary, set, or string).

### Basic Syntax

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

```

### The `range()` Function

Generates a sequence of numbers.

* `range(5)` $\rightarrow$ 0 to 4
* `range(2, 6)` $\rightarrow$ 2 to 5
* `range(1, 10, 2)` $\rightarrow$ 1, 3, 5, 7, 9 (starts at 1, ends before 10, steps by 2)

```python
for i in range(3):
    print(f"Iteration: {i}")

```

### Loop Control Statements

* **`break`**: Exits the loop immediately.
* **`continue`**: Skips the current iteration and moves to the next.
* **`pass`**: A null statement used as a placeholder (prevents errors in empty loops).
