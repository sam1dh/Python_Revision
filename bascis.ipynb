### Python Revision: Basics & Syntax

---

## 1. Comments

Used for documentation and code clarity.

* **Single-line:** Use the `#` symbol.
* **Multi-line:** Use triple quotes `"""` or `'''`. (Note: These are technically string literals ignored by Python if not assigned to a variable).

```python
# This is a single-line comment

"""
This is a multi-line 
comment or docstring.
"""

```

---

## 2. Variables & Data Types

Variables are containers; Python is **dynamically typed** (types are determined at runtime).

| Category | Type | Example |
| --- | --- | --- |
| **Numeric** | `int`, `float`, `complex` | `a = 5`, `b = 2.5`, `c = 1j` |
| **Text** | `str` | `name = "Python"` |
| **Boolean** | `bool` | `is_valid = True` |
| **Sequence** | `list`, `tuple`, `range` | `[1, 2]`, `(1, 2)`, `range(5)` |
| **Mapping** | `dict` | `{"key": "value"}` |
| **Set** | `set` | `{1, 2, 3}` |

---

## 3. Operators

### Arithmetic

* `+`, `-`, `*`, `/` : Basic Math
* `//` : **Floor Division** (returns integer) $\rightarrow$ `5 // 2 = 2`
* `%` : **Modulus** (remainder) $\rightarrow$ `5 % 2 = 1`
* `` : **Exponent** $\rightarrow$ `2  3 = 8`

### Comparison

* `==` : Equal to
* `!=` : Not equal to
* `>` , `<` , `>=` , `<=` : Relational

### Logical

* `and` : Returns True if both statements are true.
* `or` : Returns True if one statement is true.
* `not` : Reverses the logical state.

### Assignment

* `=` : Assign
* `+=` , `-=` , `*=` : **Shorthand** (e.g., `x += 3` is `x = x + 3`)

---

## 4. Quick Reference Functions

* `type(x)` : Returns the data type.
* `id(x)` : Returns the memory address.
* `input()` : Always returns data as a **string**.
* **Casting:** `int()`, `float()`, `str()`, `list()`.
