### Python Functions: Fast Revision

Functions are self-contained blocks of code designed to perform a specific task. They make your code modular, readable, and reusable (**DRY** principle: Don't Repeat Yourself).

---

## 1. Defining and Calling

* **`def` keyword**: Used to start the function definition.
* **Indentation**: The code block belonging to the function must be indented.
* **Call**: Use the name followed by parentheses `()`.

```python
def welcome_msg():
    print("Hello! Welcome to the Python Revision.")

welcome_msg() # Calling the function

```

---

## 2. Arguments vs. Parameters

* **Parameters**: The variables listed in the function definition (the placeholders).
* **Arguments**: The actual values passed to the function when calling it.

### Types of Arguments:

1. **Positional**: Passed in the order they are defined.
2. **Keyword**: Specified by name (e.g., `greet(name="Mayank")`).
3. **Default**: Parameters that have a value if none is provided.
```python
def power(base, exp=2): # exp defaults to 2
    return base ** exp

```


4. **Arbitrary (`*args`)**: Receives a **tuple** of arguments (used when the number of inputs is unknown).
5. **Arbitrary Keyword (`kwargs`)**: Receives a **dictionary** of keyword arguments.

---

## 3. The `return` Statement

A function stops executing when it hits a `return` statement. If no `return` is used, the function returns `None` by default.

```python
def multiply(a, b):
    return a * b  # Sends the result back to the caller

val = multiply(10, 5) # val is now 50

```

---

## 4. Scope (Local vs. Global)

The scope determines where a variable can be seen or accessed.

* **Local Scope**: Variables defined inside a function. They exist only while the function is running.
* **Global Scope**: Variables defined outside any function. They can be read anywhere.
* **`global` keyword**: Used inside a function to modify a variable that exists in the global scope.

---

## 5. Lambda Functions

Anonymous, one-line functions defined without the `def` keyword.

* **Syntax**: `lambda arguments : expression`
* **Use case**: Short operations, often used as arguments for functions like `map()`, `filter()`, or `sort()`.

```python
# A lambda function to double a number
double = lambda x: x * 2
print(double(5)) # 10

```

---

## 6. Docstrings

Always document your functions using triple quotes right below the definition line. It helps others (and your future self) understand the purpose of the code.

```python
def square(n):
    """Takes a number n and returns its square."""
    return n * n

```
